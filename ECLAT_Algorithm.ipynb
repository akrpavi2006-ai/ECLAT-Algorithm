```python
import pandas as pd
from pyECLAT import ECLAT
```

```python
df = pd.read_csv("/content/drive/MyDrive/AI/Market_Basket_Optimisation.csv", header=None)
```

```python
eclat = ECLAT(data=df)


indices, support = eclat.fit(min_support=0.03,
                             min_combination=2,
                             max_combination=3)
```

```python
result = pd.DataFrame({
    "Itemset": list(support.keys()),
    "Support": list(support.values())
})

result = result.sort_values(by="Support", ascending=False)

result.head(20)
```
